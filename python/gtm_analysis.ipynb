## Data Cleaning

```python
df=df.drop(columns = [
    'B2B Software Companies',
    'Create object',
    'Use AI Summary',
    'Account Priority Rationale (2)',
    'GTM Tech Stack (2)',
    'Use AI Gtm Maturity',
    'GTM Tech Stack',
    'Use AI Tier',
    'Find jobs'
],
 errors='ignore')

df.head()
```

```python
df["GTM_Jobs_Numeric"] = pd.to_numeric(
    df["Totaljobsfound"],
    errors="coerce"
)
```

## Data Analysis

```python
df[
    (df["GTM Maturity Level"] == "High") &
    (df["GTM Hiring Signal"].isin(["Low", "No observed Hiring"]))
]
```

## Data Visualization

### Number of accounts by Account Priority

```python
df['Account Priority'].value_counts().plot(kind='barh', color='green')

plt.title('Number of Accounts by Account Priority')
plt.xlabel('Number of Accounts')
plt.ylabel('Account Priority')
plt.gca().invert_yaxis()
plt.show()
```

## Python Analysis Results

```python
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('[^a-zA-Z0-9_]', '', regex=True)

df.to_csv('cleaned_target_accounts.csv', index=False)
```
